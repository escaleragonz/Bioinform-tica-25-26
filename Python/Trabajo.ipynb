#### 6.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)
1. Enzimas que tienen sitios de corte en la secuencia y posición (o posiciones de corte)
2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI' : 'G*GATCC',
    'AluI' : 'AG*CT',
    'Sau3AI' : '*GATC',
    'EcoRI' : 'G*AATTC'
}

sitios_corte = {}

for enzima, patron in enzR.items():
    partes = patron.split('*')
    sitio = ''.join(partes)
    corte_relativo = len(partes[0])
    posiciones = []
    i = secuencia.find(sitio)
    while i != -1:
        posiciones.append(i + corte_relativo)
        i = secuencia.find(sitio, i + 1)
    if posiciones:
        sitios_corte[enzima] = posiciones

print("1 Enzimas que cortan y posiciones:")
for e, pos in sitios_corte.items():
    print(f"  {e}: {pos[:10]}{'... ' if len(pos) > 10 else ''}")

sin_corte = [e for e in enzR if e not in sitios_corte]
print("2 Enzimas que NO tienen sitios de corte:", sin_corte)

cortes = sorted([p for lista in sitios_corte.values() for p in lista])

fragmentos = []
if cortes:
    inicio = 0
    for corte in cortes:
        fragmentos.append(secuencia[inicio:corte])
        inicio = corte
    fragmentos.append(secuencia[inicio:])
else:
    fragmentos = [secuencia]

print("3 Número de fragmentos generados:", len(fragmentos))
print("Mostrando solo los primeros 5 fragmentos (por tamaño):")
for i, f in enumerate(fragmentos[:5]):
    print(f"  Fragmento {i+1}: {len(f)} pb")

tamaños = [len(f) for f in fragmentos]

print("Número total de fragmentos:", len(fragmentos))
print("Tamaño de cada fragmento:")
for i, f in enumerate(fragmentos, start=1):
    print(f"  Fragmento {i}: {len(f)} pb")
```

#### 7.- Realiza una función que mediante la introducción de una secuencia de ADN devuelva lo siguiente: (3pto)
1. Todos los ORFs posibles que puedan surgir de la secuencia.
2. Selecciona los 6 ORFs más largos e indica si corresponden a la strand positiva o negativa.
3. Devuelve como mejor predicción, el más largo de todos.
4. Calcula el porcentaje de aminoácidos aromáticos de esa proteína elegida y su longitud relativa a la secuencia original de ADN.

```python
def encontrar_ORFs(secuencia):
    secuencia = secuencia.upper()
    codones_parada = ['TAA','TAG','TGA']
    codon_inicio = 'ATG'

    def ORFs_strand(seq, strand_label):
        orfs = []
        for frame in range(3):
            i = frame
            while i < len(seq) - 2:
                codon = seq[i:i+3]
                if codon == codon_inicio:
                    j = i
                    while j < len(seq)-2:
                        codon_j = seq[j:j+3]
                        if codon_j in codones_parada:
                            orf_seq = seq[i:j+3]
                            orfs.append({'seq': orf_seq, 'strand': strand_label, 'frame': frame})
                            break
                        j += 3
                i += 3
        return orfs


    orfs_pos = ORFs_strand(secuencia, 'positivo')


    complement = str.maketrans('ATGC','TACG')
    sec_revcomp = secuencia.translate(complement)[::-1]
    orfs_neg = ORFs_strand(sec_revcomp, 'negativo')

    todos_orfs = orfs_pos + orfs_neg
    todos_orfs.sort(key=lambda x: len(x['seq']), reverse=True)

    return todos_orfs

todos_orfs = encontrar_ORFs(secuencia)


print("1 Todos los ORFs posibles:", len(todos_orfs))


top6_orfs = todos_orfs[:6]

print("2 Los 6 ORFs más largos:")
for i, orf in enumerate(top6_orfs,1):
    print(f"{i}. Longitud: {len(orf['seq'])} pb, Strand: {orf['strand']}")

mejor_orf = todos_orfs[0]

print("3 Mejor predicción (ORF más largo):")
print("Longitud:", len(mejor_orf['seq']), "pb")
print("Strand:", mejor_orf['strand'])
print("Secuencia ORF:", mejor_orf['seq'])

codon_tabla = {
    'TTT':'F','TTC':'F','TTA':'L','TTG':'L',
    'CTT':'L','CTC':'L','CTA':'L','CTG':'L',
    'ATT':'I','ATC':'I','ATA':'I','ATG':'M',
    'GTT':'V','GTC':'V','GTA':'V','GTG':'V',
    'TCT':'S','TCC':'S','TCA':'S','TCG':'S',
    'CCT':'P','CCC':'P','CCA':'P','CCG':'P',
    'ACT':'T','ACC':'T','ACA':'T','ACG':'T',
    'GCT':'A','GCC':'A','GCA':'A','GCG':'A',
    'TAT':'Y','TAC':'Y','TAA':'*','TAG':'*',
    'CAT':'H','CAC':'H','CAA':'Q','CAG':'Q',
    'AAT':'N','AAC':'N','AAA':'K','AAG':'K',
    'GAT':'D','GAC':'D','GAA':'E','GAG':'E',
    'TGT':'C','TGC':'C','TGA':'*','TGG':'W',
    'CGT':'R','CGC':'R','CGA':'R','CGG':'R',
    'AGT':'S','AGC':'S','AGA':'R','AGG':'R',
    'GGT':'G','GGC':'G','GGA':'G','GGG':'G'
}

def traducir(orf_seq):
    proteina = ''
    for i in range(0,len(orf_seq)-2,3):
        codon = orf_seq[i:i+3]
        aa = codon_tabla.get(codon,'X')
        if aa == '*':
            break
        proteina += aa
    return proteina

proteina_mejor = traducir(mejor_orf['seq'])

aromaticos = sum(1 for aa in proteina_mejor if aa in ['F','Y','W'])
porcentaje_aromaticos = aromaticos / len(proteina_mejor) * 100 if len(proteina_mejor)>0 else 0

longitud_relativa = len(proteina_mejor)*3 / len(secuencia) * 100

print("4 Proteína traducida:", proteina_mejor)
print("Porcentaje aminoácidos aromáticos:", round(porcentaje_aromaticos,2), "%")
print("Longitud relativa a secuencia ADN:", round(longitud_relativa,2), "%")
```
